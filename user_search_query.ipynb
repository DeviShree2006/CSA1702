```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# -----------------------------
# Sample Web Pages (Documents)
# -----------------------------
documents = [
    "Python is a programming language used for AI and Machine Learning.",
    "Docker is a containerization platform used for deploying applications.",
    "Convolutional Neural Networks are used in image classification.",
    "Machine Learning includes supervised and unsupervised learning.",
    "Deep Learning is a subset of Machine Learning."
]

# -----------------------------
# User Search Query
# -----------------------------
query = input("Enter your search query: ")

# -----------------------------
# Convert text into TF-IDF vectors
# -----------------------------
vectorizer = TfidfVectorizer()

all_text = documents + [query]

tfidf_matrix = vectorizer.fit_transform(all_text)

# -----------------------------
# Calculate Similarity
# -----------------------------
query_vector = tfidf_matrix[-1]

document_vectors = tfidf_matrix[:-1]

similarity = cosine_similarity(query_vector, document_vectors)

# -----------------------------
# Rank Documents
# -----------------------------
scores = similarity.flatten()

ranked = sorted(enumerate(scores),
                key=lambda x: x[1],
                reverse=True)

# -----------------------------
# Display Results
# -----------------------------
print("\nTop Search Results\n")

for index, score in ranked:
    print(f"Score: {score:.3f}")
    print(documents[index])
    print("-"*60)
```
