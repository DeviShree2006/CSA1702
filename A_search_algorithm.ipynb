```python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1), ('G', 2)],
    'D': [],
    'E': [],
    'F': [],
    'G': [('Goal', 1)],
    'Goal': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 3,
    'D': 8,
    'E': 7,
    'F': 4,
    'G': 1,
    'Goal': 0
}

def astar(start, goal):
    pq = [(heuristic[start], 0, start)]

    while pq:
        f, g, node = heapq.heappop(pq)

        print(node, end=" ")

        if node == goal:
            print("\nGoal Found!")
            return

        for neighbor, cost in graph[node]:
            heapq.heappush(
                pq,
                (g + cost + heuristic[neighbor], g + cost, neighbor)
            )

astar('A', 'Goal')
```
