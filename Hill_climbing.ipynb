```python
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 1), ('G', 2)],
    'D': [],
    'E': [],
    'F': [],
    'G': [('Goal', 1)],
    'Goal': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 3,
    'D': 8,
    'E': 7,
    'F': 4,
    'G': 1,
    'Goal': 0
}

def hill_climbing(start, goal):
    current = start

    while True:
        print(current, end=" ")

        if current == goal:
            print("\nGoal Found!")
            return

        neighbors = graph[current]

        if not neighbors:
            print("\nStuck!")
            return

        best = min(neighbors, key=lambda x: heuristic[x[0]])

        if heuristic[best[0]] >= heuristic[current]:
            print("\nLocal Maximum Reached!")
            return

        current = best[0]

hill_climbing('A', 'Goal')
```
