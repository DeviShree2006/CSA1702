```python
# Monkey Banana Problem

monkey = "door"
box = "window"
banana = "ceiling"

# Initial state
print("Initial State:")
print("Monkey is at", monkey)
print("Box is at", box)
print("Banana is at", banana)

# Step 1: Monkey moves to the box
monkey = box
print("\n1. Monkey moves to the box.")
print("Monkey is at", monkey)

# Step 2: Monkey pushes the box below the banana
box = banana
monkey = banana
print("\n2. Monkey pushes the box below the banana.")
print("Monkey is at", monkey)
print("Box is at", box)

# Step 3: Monkey climbs the box
print("\n3. Monkey climbs the box.")

# Step 4: Monkey gets the banana
print("\n4. Monkey gets the banana!")
print("Goal achieved.")
```
