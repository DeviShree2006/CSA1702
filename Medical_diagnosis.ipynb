```python
# Medical Diagnosis System

print("Medical Diagnosis System")
print("------------------------")

fever = input("Do you have fever? (yes/no): ").lower()
cough = input("Do you have cough? (yes/no): ").lower()
headache = input("Do you have headache? (yes/no): ").lower()
body_pain = input("Do you have body pain? (yes/no): ").lower()
sore_throat = input("Do you have sore throat? (yes/no): ").lower()


# Rules for diagnosis

if fever == "yes" and cough == "yes" and body_pain == "yes":
    print("\nPossible diagnosis: Flu")

elif fever == "yes" and sore_throat == "yes" and cough == "yes":
    print("\nPossible diagnosis: Throat infection")

elif headache == "yes" and fever == "yes":
    print("\nPossible diagnosis: Fever-related illness")

elif cough == "yes" and sore_throat == "yes":
    print("\nPossible diagnosis: Common cold")

else:
    print("\nDiagnosis could not be determined.")
    print("Please consult a qualified medical professional.")
```
