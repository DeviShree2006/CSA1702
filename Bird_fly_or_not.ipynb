```python
# List of birds
birds = ["parrot", "eagle", "penguin", "ostrich"]

# Birds that can fly
can_fly = ["parrot", "eagle"]

# Ask the user for a bird
bird = input("Enter a bird: ").lower()

# Check whether the bird is known
if bird in birds:

    if bird in can_fly:
        print(bird, "can fly.")
    else:
        print(bird, "cannot fly.")

else:
    print("Bird not found.")
```
